# Chapter 2 – Why RAG Exists

Imagine a 500-page PDF.

Large Language Models have limited context windows (token limits). They cannot always use every page at once.

Without relevant context they may generate incorrect answers (hallucinations).

RAG solves this by retrieving only the most relevant pieces before asking the LLM.

```
User Question
      |
      v
Relevant Pages
      |
      v
LLM
      |
      v
Answer
```

## Stage Summary

| Stage | Purpose |
|---|---|
|Read File|Load document|
|Preprocess|Clean text|
|Chunk|Split into small parts|
|Embedding|Convert meaning into vectors|
|Vector Database|Store vectors efficiently|
|Retrieve|Find relevant chunks|
|Prompt Construction|Combine question and retrieved text|
|LLM|Generate answer|


````python
pipeline = [
    "Read File",
    "Preprocess",
    "Chunk",
    "Embedding",
    "Vector Database",
    "Retrieve",
    "Prompt Construction",
    "LLM",
    "Answer"
]

for step in pipeline:
    print(step)
````
