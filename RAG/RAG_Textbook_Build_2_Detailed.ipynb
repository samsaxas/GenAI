## Chapter 4 – File Handling
### What is a File?
A file is a collection of data stored permanently on a storage device.

## Problem
Programs need a way to save information after they stop running.

## Why do we need this?
Without files, every program would lose its data when it closes.

## Real-life Analogy
A notebook stores notes permanently. A file stores digital information permanently.

## Theory
A file is made of bytes. Python reads bytes and converts them into text using an encoding.

## What is Encoding?
Encoding is a rule that converts characters to bytes and back.

## UTF-8
UTF-8 is the world's most common encoding. It supports English and most other languages.

## Why open()?
`open()` asks the operating system to give Python access to a file.

## Why with open()?
It automatically closes the file even if an error occurs.

## Internal Working
```
Disk -> Operating System -> Python open() -> File Object -> read() -> String in RAM
```
Python creates a file object, requests bytes, decodes them using UTF-8, and stores the resulting string in memory.

### Code Example
The following code creates a file and reads it back.

````
with open("sample.txt","w",encoding="utf-8") as f:
    f.write("RAG starts with reading documents.")

with open("sample.txt","r",encoding="utf-8") as f:
    text=f.read()

print(text)
````

### Code Explanation
- `open()` creates a file object.
- Mode `w` writes data.
- Mode `r` reads data.
- `encoding='utf-8'` tells Python how to decode characters.
- `f.read()` returns a string.

### Memory Representation
`f` -> File object
`text` -> Python string object in RAM

# Chapter 5 – Text Preprocessing
Text preprocessing prepares raw text for retrieval.

## Lowercasing
- Problem: `Cat` and `cat` look different.
- Why: Make comparisons consistent.
- Input: `Cat`
- Output: `cat`
- Advantage: Better matching.
- Disadvantage: Loses capitalization.
- If skipped: Same word may be treated differently.

## Remove Punctuation
- Problem: Symbols add noise.
- Why: Improve matching.
- Input: `hello!!!`
- Output: `hello`
- Disadvantage: Some punctuation carries meaning.

## Remove Extra Spaces
- Problem: Inconsistent formatting.
- Output: Clean text.

## Tokenization
- Problem: Computer cannot directly analyze long sentences.
- Output: List of words.

## Stopword Removal
- Removes very common words when appropriate.
- Skipping keeps more context but increases size.

## Lemmatization
- Converts words to dictionary form.
- Example: `running -> run`.


### Code Example

````
import re

text = "Cats, ARE running!!!"

clean = text.lower()
clean = re.sub(r"[^a-zA-Z\s]", "", clean)
tokens = clean.split()

print(clean)
print(tokens)
````

### Code Explanation
Every preprocessing step transforms the previous output into cleaner text. The final list of tokens is easier for later stages such as chunking and embeddings.

# Chapter 6 – Chunking
## Problem
Large Language Models cannot process unlimited text.

## Token Limits
Every model has a maximum context window. If the document is too large, only part of it can be processed.

## Why Chunk?
Split one large document into many smaller pieces.

## Why Overlap?
Sentences near chunk boundaries keep shared context.

## Why Recursive Splitting?
It first tries large separators (paragraphs), then smaller ones (lines, spaces, characters) to preserve meaning.

## Why 500 Characters?
It is a practical balance between context and retrieval speed. Different applications may choose different sizes.

## Why Not the Whole Document?
Whole documents waste context window, reduce retrieval precision, and may exceed token limits.

### RecursiveCharacterTextSplitter
1. Try splitting by paragraphs.
2. If chunk is still too large, split by newline.
3. Then spaces.
4. Finally characters.

```
Document
 |
 +--Chunk 1----+
        overlap
          +--Chunk 2----+
                 overlap
                    +--Chunk 3
```

### Simple Chunking Example

````
text = "ABCDEFGHIJKLMNOPQRSTUVWXYZ" * 8

chunk_size = 40
overlap = 10

chunks = []
start = 0

while start < len(text):
    end = start + chunk_size
    chunks.append(text[start:end])
    start += chunk_size - overlap

for i, c in enumerate(chunks, 1):
    print(f"Chunk {i}: length={len(c)}")
````
