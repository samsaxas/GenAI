```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
```

STEP 1: Loading the csv file

```python
df = pd.read_csv('spam.csv', encoding = 'latin-1')  # load the data
```

```python
df.head()
```

In RAG, we do not segregate input and output.

```python
x = df['v2']
y = df['v1']
```

### Vectorization
#### Since the model cannot understand text, it will only understand numbers, we need to use vectorization

```python
vectors = CountVectorizer(stop_words='english') # acts as embedding model
```

```python
model = KNeighborsClassifier() # acts as vector database, stores the vectors
# model.fit(x,y)  # will give error
# ValueError: Expected a 2-dimensional container but got <class 'pandas.Series'> instead. Pass a DataFrame containing a single row (i.e. single sample) or a single column (i.e. single feature) instead.
```

```python
vectors.fit_transform(x) # returns sparse matrix
values = vectors.fit_transform(x).toarray()  # converted sentences into vectors

cols = vectors.get_feature_names_out()

updated_x = pd.DataFrame(values, columns = cols)
updated_x
```

```python
model.fit(updated_x, y)  # stored the information (input and output)(stores the row vectors)
```

```python
new_input = [df['v2'][0]]   # here it is not in numbers, the trained model having numbers is comparing with text in new input
new_input  # this is first row
```

```python
# model.predict(new_input)
```

```python
updated_new_input = vectors.transform(new_input).toarray()
# updated_new_cols = vectors.get_feature_names_out()
```

```python
model.predict(updated_new_input)
```
