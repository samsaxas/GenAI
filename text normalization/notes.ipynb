# Retrieval-Augmented Generation (RAG)

RAG is used to obtain **accurate results** by retrieving relevant information from external documents before generating the final response.

## RAG Workflow

### Step 1: LangChain

The first step is to upload the documents.

**Supported file formats:**
- `.txt`
- `.pdf`
- `.doc`

↓

### Step 2: Text Cleaning (CNLP)

The uploaded text is cleaned using NLP libraries.

**Libraries used:**
- NLTK
- SpaCy

↓

### Step 3: Tokens & Types

The cleaned text is converted into **tokens** and token types.

↓

### Step 4: Chunking & Types (Google Gemini)

The text is divided into smaller chunks.

- Chunking
- Types

↓

### Step 5: Embeddings (Gemini)

The text is converted into vectors (numbers).

↓

### Step 6: Vector Database

The generated vectors are stored inside a **Vector Database**.

**Example:**
- FAISS (Facebook AI)

↓

### Step 7: Similarity Search (SS)

The vector database performs similarity search.

Two common techniques:

#### KNN (K-Nearest Neighbour)

- Finds the nearest neighbours.

#### Cosine Similarity

- Statistical concept.
- Used to find similar content.

---

**Goal:** To get accurate results.

## Tokenization

Why tokenization is required?
- NLP is Natural Language Processing comes under AI
- NLP helps machine to understand the human language
- NLTK and Spacy
- Model fails to understand the context, so we need to convert into tokens

- Splitting the data into small pieces


- If the data is in tokens, machine can easily understand it and the accuracy will increase

### Types of Tokenization
1. Character Level Tokenization
 - Converts data.txt -> characters
 - We cannot use this in LLM -> RAG because in 1 prompt it will reach the limit and the prediction will be wrong.
 - autocorrect 

example: "ML is Superset of DL"

USE: encryption decryption

2. Word Level Tokenization
- Converts the data into words
- USES: Spam detection

3. Sentence Level Tokenization
- USES: Sentiment Analysis and translation

4. Sub Word Tokenization
- Converts data to words
- USES: LLM -> RAG
- example: Tokenization -> Token + zation
- uses mistral, llama, gpt
- claude


NLTK and Spacy
- Stop Words Removal
- autocorrect and pdf extraction

### CHUNKING is nothing but splitting the document to smaller piece.
- Chunks are nothing but group of words or characters.

Types of chunking
1. Fixed size chunking -> `chunks divided into fixed characters`
2. Fixed size overlapping -> `used for RAG in small projects`
3. Paragraph based -> `inside recursive chunking`
4. Sentence based -> `inside recursive chunking`
5. Recursive -> `used for real-time RAG based projects`
6. Semantic -> `for large documents`
7. Page Number or Structured Chunking
8. Standard Chunking

### Chunking types used in RAG Architecture
- Fixed size + overlapping 
- Recursive
- Structure based
- Semantic Chunking


1. Fixed size + overlapping -> all the chunks are of equal size, but except the first chunk, in other chunks, the chunk contains previous chunk's part according to overlappiing size.
2. Recursive -> includes multiple types of other chunking, it finds the best chunking according to the scenario
3. Structure based -> it chunks based on page number
4. Semantic Chunking -> used for large dataset.
