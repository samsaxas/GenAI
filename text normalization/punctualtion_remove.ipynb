```python
# to remove the punctuations from data.txt file using re.sub
import re
with open('data.txt', 'r+') as file:
    data = file.read()
    data_no_punct = re.sub(r'[^\w\s]', '', data)
    file.seek(0)
    file.write(data_no_punct)
print(data_no_punct)
```

```python
# use re (sub) maketrans and translate
import string
with open('data.txt', 'r+') as file:
    data = file.read()
    table = str.maketrans('', '', string.punctuation)
    data_no_punct = data.translate(table)
    file.seek(0)
    file.write(data_no_punct)
print(data_no_punct)
```
