```python
nlp_sentence = ["NLP stands for Natural Language Processing.",
"NLP enables computers to understand human language.",
"NLP combines linguistics with artificial intelligence.",
"NLP is used in chatbots and virtual assistants.",
"NLP helps analyze text and speech data.",
"NLP powers machine translation systems.",
"NLP is used for sentiment analysis.",
"NLP can summarize large amounts of text.",
"NLP improves communication between humans and machines.",
"NLP has applications in healthcare, education, and business."]
# create 10 sentences for nlp
```

```python
# assignment
# sentence level + stop word removal
```

```python
text = " ".join(nlp_sentence)
sent_tokens = sent_tokenize(text)

sentence_level_tokens = []
for sent in sent_tokens:
    word_tokens = word_tokenize(sent)
    filtered_words = [w for w in word_tokens if w.lower() not in stop_words]
    sentence_level_tokens.append(filtered_words)

sentence_level_tokens
```
