```python
import nltk
nltk.download("wordnet")
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
tokens = [
    "running",
    "ran",
    "cars",
    "geese",
    "better",
    "children",
    "was",
    "studies",
    "happier",
    "leaves"
]

for word in tokens:
    print(word, lemmatizer.lemmatize(word))
```

```python
for word in tokens:
    print(word, lemmatizer.lemmatize(word, pos='v'))
```

BY Default it uses Part Of Speed -> Noun(N)
- lemmatize skips for remaining words

PARAMETERS

word : str
The input word to lemmatize.

pos : str
The Part Of Speech tag. Valid options are "n" for nouns, "v" for verbs, "a" for adjectives, "r" for adverbs and "s" for satellite adjectives.

- Here, pos will take only 1 part of speech, rest all words will be ignored. THAT'S LIMITATION

| POS Tag  | Meaning                                                                                                                                  | Example                             |
| -------- | ---------------------------------------------------------------------------------------------------------------------------------------- | ----------------------------------- |
| **NN**   | **Noun (Singular)** – A naming word referring to one person, place, thing, or idea.                                                      | `dog`, `book`, `car`                |
| **NNS**  | **Noun (Plural)** – A naming word referring to more than one person, place, thing, or idea.                                              | `dogs`, `books`, `cars`             |
| **NNP**  | **Proper Noun (Singular)** – The specific name of one person, place, organization, or thing.                                             | `India`, `Samriddhi`, `Google`      |
| **NNPS** | **Proper Noun (Plural)** – The specific names of more than one person, place, or organization.                                           | `Americans`, `Saxenas`, `Himalayas` |
| **VB**   | **Verb (Base Form)** – The original form of a verb without any tense or ending.                                                          | `go`, `eat`, `run`                  |
| **VBD**  | **Verb (Past Tense)** – A verb showing an action that happened in the past.                                                              | `went`, `ate`, `ran`                |
| **VBG**  | **Verb (Gerund / Present Participle)** – A verb ending in **-ing**, used as a noun or to form continuous tenses.                         | `running`, `eating`, `playing`      |
| **VBN**  | **Verb (Past Participle)** – A verb form used with **has**, **have**, **had**, **is**, **was**, etc., to form perfect or passive tenses. | `gone`, `eaten`, `written`          |
| **VBP**  | **Verb (Non-3rd Person Singular Present)** – Present tense verb used with **I, you, we, they**.                                          | `I run`, `They eat`                 |
| **VBZ**  | **Verb (3rd Person Singular Present)** – Present tense verb used with **he, she, it**, usually ending in **-s** or **-es**.              | `He runs`, `She eats`               |
| **JJ**   | **Adjective** – A word that describes or modifies a noun.                                                                                | `beautiful`, `big`, `happy`         |
| **JJR**  | **Comparative Adjective** – An adjective used to compare two things.                                                                     | `bigger`, `faster`, `smarter`       |
| **JJS**  | **Superlative Adjective** – An adjective showing the highest degree among three or more things.                                          | `biggest`, `fastest`, `smartest`    |
| **RB**   | **Adverb** – A word that describes a verb, adjective, or another adverb.                                                                 | `quickly`, `very`, `carefully`      |
| **RBR**  | **Comparative Adverb** – An adverb used to compare two actions.                                                                          | `faster`, `more carefully`          |
| **RBS**  | **Superlative Adverb** – An adverb showing the highest degree among three or more actions.                                               | `fastest`, `most carefully`         |


N = 'n'
V = 'v'
JJ = 'a' adjectives
RBR = 'r' adverbs

```python
def find_postag(tag):
    if tag.startswith('J'):
        return 'a'
    elif tag.startswith('N'):
        return 'n'
    elif tag.startswith('V'):
        return 'v'
    elif tag.startswith('R'):
        return 'r'
    else:
        return 'n'

```
