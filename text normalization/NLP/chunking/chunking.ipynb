## CHUNKING
Coverting large data into smaller parts.

## Where we use chunking?
We use chunking before embedding.

## TYPES OF CHUNKING
1. Fixed size chunking -> `chunks divided into fixed characters`
2. Fixed size overlapping -> `used for RAG in small projects`
3. Paragraph based -> `inside recursive chunking`
4. Sentence based -> `inside recursive chunking`
5. Recursive -> `used for real-time RAG based projects`
6. Semantic -> `for large documents`
7. Page Number or Structured Chunking
8. Standard Chunking

### FIxed Size and Overlapping -> Used in RAG
- has 2 parameters -> fixed size and overlapping
- second chunk starts from end of first chunk, likewise....
- it preserves the meaning of the word by overlapping

### How we can decide fixed_size and overlapping?
- Overlapping should be 15% - 20% of fixed_size.
- Not 50%, because it will give repeated character and requires more memory instead of preserving the meaning of the context.

```python
with open('data.txt') as fh:
    data = fh.read()
print(data)
```

```python
size = 20
overlap = 4
chunks = []
# for char in data:
#     chunks = data[:size]
for index in range(0,len(data),size):
    print(index, end = ' ')
    if index > 0:
    # chunks += data[index:index+size] # data[0:0+20]
        chunk = (data[index-overlap:index+size])
    else:
        chunk = data[index:index+size]
    chunks.append(chunk)

    
chunks
```
