This notebook demonstrates a complete document preprocessing pipeline used in Natural Language Processing (NLP). The pipeline includes file handling, text normalization, spelling correction, lemmatization, and recursive chunking before preparing the text for downstream tasks such as Retrieval-Augmented Generation (RAG), Semantic Search, Sentiment Analysis, and Large Language Models (LLMs).

# Step 1: Import Required Libraries

Import all the libraries required throughout the preprocessing pipeline.

````python
import re
import string 
import emoji
import contractions
import spacy
from nltk.tokenize import word_tokenize
from langchain_text_splitters import RecursiveCharacterTextSplitter
from textblob import TextBlob
````

# Step 2: File Handling

Read the text file using UTF-8 encoding.

UTF-8 allows Python to correctly read Unicode characters such as emojis.

The file is opened using read mode (`r`), and its contents are stored in the variable `text`.

````python
with open('tn_data.txt','r', encoding = 'utf-8') as file:
    text = file.read()
````

while creating dataframe, file path, encoding, chunk_size are the parameters

# Step 3: TEXT NORMALIZATION

# Step 3.1: Convert Text to Lowercase

Lowercase conversion converts all characters to lowercase so that words with different capitalizations are treated as the same word.

Example:

Python

PYTHON

python

becomes

python

````python
text = text.lower()
````

````python
text
````

# Step 3.2: Remove Extra Spaces

Text cleaning often creates multiple consecutive spaces.

Replace them with a single space.

````python
# text.replace('  ', ' ') # this will only replace when there are 2 spaces
text = re.sub(r'\s{2,}',' ',text) # replaces 2 or more spaces with 1 space
text
````

# Step 3.3: Expand Contractions & Abbreviations

Contractions shorten words in English.

Examples:

I'm → I am

can't → cannot

it's → it is

ASAP → As Soon As Possible

Expanding contractions improves consistency during NLP preprocessing.

````python
text = contractions.fix(text)
text
````

# Step 3.4: Remove Punctuation

Punctuation marks generally do not contribute meaningful information for most NLP tasks.

Python provides the `string.punctuation` constant to identify punctuation symbols.

````python
string.punctuation
````

````python
punc = string.punctuation
text = re.sub(f'[{punc}]','',text)
text
````

# Step 3.5: Handle Emojis

Emoji preprocessing depends on the NLP task.

### General NLP Tasks

For document preprocessing, semantic search, embeddings, RAG, and chunking, emojis usually do not add useful information.

Therefore, they can be removed using `replace_emoji()`.

### Sentiment Analysis

When performing sentiment analysis, emojis represent emotions.

Instead of removing them, convert them into descriptive text using `demojize()`.

````python
text = emoji.replace_emoji(text)
text
````

# Step 3.6: Remove Special Characters

Remove symbols while keeping letters, digits, and spaces.

The `isalnum()` method checks whether a character is alphanumeric.

Use demojize or replace_emoji before removing special characters. Because, while removing special characters, it removes emojis also.

Also, removing special characters will remove extra spaces.

````python
# import warnings
text = re.sub(r'\W+',' ',text) # or use r string to remove warning
# re.sub('[^0-9a-zA-Z\s]','', data)
# warnings.filterwarnings('ignore')
text
````

# Step 3.7: Correct Spelling using TextBlob

TextBlob automatically corrects many common spelling mistakes.

Example:

machine learning

````python
# 1.
text = str(TextBlob(text).correct()) # convert to str because it will return TextBlob object
# 2.
# TextBlob(text).correct().raw_sentences
# 3.
# ' '.join(text)
text
````

# Step 4: Tokenization & Lemmatization

Tokenization splits text into individual words called tokens.

These tokens can later be converted back into a paragraph after preprocessing.

Lemmatization converts words into their dictionary (base) form while preserving their meaning.

Examples

running → run

studies → study

children → child

went → go

````python
nlp = spacy.load('en_core_web_sm')
````

````python
tokens = nlp(text)
updated_tokens = [token.lemma_ for token in tokens if not token.is_stop]
data = ' '.join(updated_tokens)
data
````

# Step 5: Recursive Chunking

Large Language Models have a limited context window.

Recursive Character Text Splitting divides large documents into smaller overlapping chunks while preserving context.

Separator priority:

1. Paragraph

2. Line

3. Space

4. Character

````python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=20
)
chunks = splitter.split_text(data)
````
