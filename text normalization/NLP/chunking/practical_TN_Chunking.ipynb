```python
import string 
import emoji
import contractions
import spacy
from nltk.tokenize import word_tokenize
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

LOAD THE DATA

```python
with open('tn_data.txt','r', encoding = 'utf-8') as file:
    text = file.read()
```

```python
print(text)
```

CONVERTING TEXT TO LOWERCASE

```python
text = text.lower()
```

```python
text
```

REMOVE PUNCTUTATION

we must use contraction first, before removing punctuations becuase for `don't` -> `dont` the meaning will change.

```python
text = ''.join(char for char in text if char not in string.punctuation)  # it is a variable
```

```python
print(text, len(text), sep = '\n')
```

REMOVE SPECIAL CHARACTERS

```python
text = ''.join(char if char.isalnum() or char.isspace() else ' ' for char in text)
```

```python
print(text, len(text), sep = ' ')
```

HANDLE EMOJIS
- Special character will handle emojis also

```python
text = emoji.demojize(text)
```

```python
print(text, len(text), sep = '\n')
```

EXPAND CONTRACTION

```python
text = contractions.fix(text)
```

```python
print(text, len(text), sep = '\n')
```

REMOVE EXTRA SPACES

```python
text = ' '.join(text.split())
```

```python
import re
re.sub('\s{2,}', ' ',text)
```

```python
print(text, len(text), sep='\n')
```

CORRECT SPELLINGS USING TEXTBLOB

```python
from textblob import TextBlob
text = str(TextBlob(text).correct())
```

```python
print(text, len(text), sep='\n')
```

TOKENIZATION

```python
tokens = word_tokenize(text)
print(tokens)
```

COMBINE TOKENS

```python
normalized_text = ' '.join(tokens)
```

```python
normalized_text
```

LEMMATIZATION

```python
nlp = spacy.load("en_core_web_sm")
print(nlp)
```

```python
doc = nlp(normalized_text)
print(doc)
```

```python
lemmatized_text = ' '.join(token.lemma_ for token in doc)
print(lemmatized_text)
```

RECURSIVE CHUNKING

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)
chunks = splitter.split_text(lemmatized_text)
```

```python
chunks
```

PRINT CHUNKS

```python
for i, chunk in enumerate(chunks, start=1):
    print(f"Chunk {i}")
    print(chunk)
    print("-" * 80)
```
