```python
from nltk.tokenize import word_tokenize, sent_tokenize
```

| word tokenization | sentence tokenization |
|---|---|
|uses word_tokenize|uses sent_tokenize|
|breaks the text, word by word| breaks the text, sentence by sentence|
|splits using space|splits using period or dot|


```python
text = '''Return a tokenized copy of     text, using NLTK's recommended word tokenizer (currently an improved. 
TreebankWordTokenizer along with .  PunktSentenceTokenizer for the specified language).
'''
```

```python
# text = "I am Samriddhi Saxena. I am. a Student"
word_tokens = word_tokenize(text) #breaks the sentence into words
# word_tokens
```

```python
# text.split()
```

```python
sent_tokens = sent_tokenize(text) # breaks text into sentences
sent_tokens
```
