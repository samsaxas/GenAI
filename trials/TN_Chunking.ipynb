### TEXT NORMALIZATION

### 1. CONVERTING THE TEXT TO LOWERCASE

```python
data = data.lower()
data
```

### 2. REMOVING THE PUNCTUATIONS

```python
import string
punctuations = string.punctuation
punctuations
```

```python
# file.tell()
```

```python
# file.close()
# file = open('data.txt','w')
for p in punctuations:
    data = data.replace(p, '')
# file.seek(0)
# file.write(data)
data
```

### 3. REMOVING SPECIAL CHARACTERS

```python
# to remove punctuations from the data.txt
# with open('data.txt', 'r+') as file:
#     data = file.read()
for char in data:
    if not char.isalnum() and not ord(char) == 32:
        data = data.replace(char, '')
            # print(data)
            # print()
print(data)
```

### 4. HANDLING EMOJIS

```python
import emoji
data = emoji.replace_emoji(data,'')
```

```python
data
```

### 5. EXPANDING WORDS & ABBREVIATIONS

```python
import contractions
data = contractions.fix(data)
data
```

### 6. REPLACE TWO SPACES TO ONE SPACE

```python
data = data.replace('  ', ' ')
data
```

## SPACY LEMMATIZATION

```python
import spacy
```

```python
nlp = spacy.load('en_core_web_sm')
```

```python
tokens = nlp(data)
tokens
```

```python
for token in tokens:
    print(token.lemma_)
```

```python
lemmatize_tokens = [token.lemma_ for token in tokens if not token.is_stop]
lemmatize_tokens
```

```python
data = ' '.join(lemmatize_tokens)
```

```python
data
```

```python
for token in tokens:
    print(token.pos_, token.tag_)
```
