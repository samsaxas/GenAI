# 📘 Document Preprocessing using File Handling, Text Normalization, Lemmatization, and Recursive Chunking

This guide explains the complete preprocessing pipeline for preparing textual documents before using them in Natural Language Processing (NLP) applications such as Retrieval-Augmented Generation (RAG), Semantic Search, Question Answering, and Large Language Models (LLMs).

The pipeline begins by reading a text file using **file handling**, followed by **text normalization**, **lemmatization**, and finally **recursive chunking**.

---

# Workflow

```
Load Documents (File Handling)
            ↓
Read File Content
            ↓
Text Normalization
            ↓
Tokenization
            ↓
Lemmatization
            ↓
Recursive Chunking
            ↓
Print List of Chunks
```

---

# Step 1: File Handling

## Explanation

File handling is the process of reading data from or writing data to files stored on the computer.

In NLP, most datasets are stored as files such as:

- TXT
- PDF
- DOCX
- CSV

Before processing the text, it must first be loaded into Python.

Python provides the `open()` function to access files.

---

## Syntax

```python
open(file_name, mode, encoding)
```

### Parameters

- **file_name** → Name or path of the file.
- **mode** → Specifies how the file should be opened.
- **encoding** → Specifies the character encoding (commonly UTF-8).

---

## File Modes

| Mode | Description |
|------|-------------|
| `r` | Read an existing file |
| `w` | Write to a file (creates a new file or overwrites an existing one) |
| `a` | Append data to the end of a file |
| `r+` | Read and write |
| `rb` | Read a binary file |
| `wb` | Write a binary file |

For document preprocessing, we generally use **read (`r`) mode**.

---

## Python Code

```python
with open("sample.txt", "r", encoding="utf-8") as file:
    text = file.read()

print(text)
```

### Explanation

- `with open()` automatically closes the file after use.
- `"sample.txt"` is the file to be opened.
- `"r"` opens the file in read mode.
- `"utf-8"` allows reading Unicode characters such as emojis.
- `file.read()` reads the complete file into a single string.

# Step 2: Text Normalization

## Explanation

Text normalization is the process of converting raw text into a clean and consistent format before further processing.

Real-world text often contains:

- Uppercase letters
- Punctuation
- Special characters
- Emojis
- Contractions
- Extra spaces

Cleaning the text improves the quality of NLP tasks such as tokenization, embeddings, semantic search, and information retrieval.

The normalization process includes the following steps.

---

# Step 2.1: Convert Text to Lowercase

## Explanation

Words with different capitalizations represent the same meaning.

Example

```
Python
PYTHON
python
```

After conversion

```
python
```

---

## Python Code

```python
text = text.lower()
```

---

# Step 2.2: Remove Punctuation

## Explanation

Punctuation marks usually do not contribute meaningful semantic information.

Examples

```
.
,
!
?
:
;
'
"
(
)
```

Python provides the `string.punctuation` constant to identify punctuation characters.

---

## Python Code

```python
import string

text = ''.join(char for char in text if char not in string.punctuation)
```

---

# Step 2.3: Remove Special Characters

## Explanation

Special characters such as

```
@
#
$
%
&
*
^
~
|
\
```

are removed to retain only letters, numbers, and spaces.

The `isalnum()` method checks whether a character is alphanumeric.

---

## Python Code

```python
text = ''.join(char if char.isalnum() or char.isspace() else ' ' for char in text)
```

## Explanation

Lemmatization converts words to their base or dictionary form while preserving their meaning.

Examples

|--------------|-------|
| running | run |
| studies | study |
| cars | car |
| children | child |

Unlike stemming, lemmatization produces valid dictionary words.

# Step 4: Recursive Chunking

## Explanation

Large Language Models have a limited context window.

Instead of passing an entire document, the text is divided into smaller overlapping pieces called **chunks**.

Recursive Character Text Splitting attempts to split the document using increasingly smaller separators.

Separator order:

1. Paragraph (`\n\n`)
2. Line (`\n`)
3. Space (` `)
4. Character (`""`)

This helps preserve the context while creating chunks.

---

## Advantages

- Preserves sentence structure
- Maintains contextual information
- Improves semantic search
- Suitable for Retrieval-Augmented Generation (RAG)
- Prevents context window overflow

---

## Install LangChain

```python
pip install langchain
```

---

## Python Code

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = splitter.split_text(lemmatized_text)
```

---

# Step 5: Print the Chunks

## Explanation

After chunking, each chunk is stored as an element of a Python list.

These chunks are later converted into embeddings and stored in vector databases for retrieval.

---

## Python Code

```python
for i, chunk in enumerate(chunks, start=1):
    print(f"Chunk {i}")
    print(chunk)
    print("-" * 80)
```

# Complete Pipeline

```python
import string
import emoji
import contractions
import spacy

from nltk.tokenize import word_tokenize
from langchain.text_splitter import RecursiveCharacterTextSplitter

# Step 1: Read the file
with open("sample.txt", "r", encoding="utf-8") as file:
    text = file.read()

# Step 2: Convert to lowercase
text = text.lower()

# Step 3: Remove punctuation
text = ''.join(char for char in text if char not in string.punctuation)

# Step 4: Remove special characters
text = ''.join(char if char.isalnum() or char.isspace() else ' ' for char in text)

# Step 5: Handle emojis
text = emoji.demojize(text)

# Step 6: Expand contractions
text = contractions.fix(text)

# Step 7: Remove extra spaces
text = " ".join(text.split())

# Step 8: Tokenization
tokens = word_tokenize(text)

# Step 9: Combine tokens into a paragraph
normalized_text = " ".join(tokens)

# Step 10: Lemmatization
nlp = spacy.load("en_core_web_sm")
doc = nlp(normalized_text)
lemmatized_text = " ".join(token.lemma_ for token in doc)

# Step 11: Recursive chunking
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = splitter.split_text(lemmatized_text)

# Step 12: Print chunks
for i, chunk in enumerate(chunks, start=1):
    print(f"Chunk {i}")
    print(chunk)
    print("-" * 80)
```
