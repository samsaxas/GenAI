# Step 2: Text Normalization

## Explanation

Text normalization converts raw text into a clean and consistent format.

Raw documents usually contain:

- Uppercase letters
- Punctuation
- Emojis
- Extra spaces
- Contractions
- Special symbols

Cleaning the text improves the quality of tokenization, embedding generation, and information retrieval.

The normalization process consists of the following sub-steps.

---

# Step 2.1: Convert Text to Lowercase

## Explanation

Converting everything to lowercase ensures that words with different capitalization are treated as the same word.

Example:

```
Python
PYTHON
python
```

All become

```
python
```

---

## Python Code

```python
text = text.lower()
```

---

# Step 2.2: Remove Punctuation

## Explanation

Punctuation marks generally do not contribute meaningful semantic information.

Examples:

```
.
,
!
?
:
;
"
'
(
)
```

These characters are removed using the `string.punctuation` constant.

---

## Python Code

```python
import string

text = ''.join(char for char in text if char not in string.punctuation)
```

---


# Step 2.3: Remove Special Characters

## Explanation

Special characters such as:

```
@
#
$
%
&
*
^
~
|
\
```

are removed to retain only letters and numbers.

The `isalnum()` method checks whether a character is alphanumeric.

---

## Python Code

```python
text = ''.join(char if char.isalnum() or char.isspace() else ' ' for char in text)
```

---

# Step 2.4: Handle Emojis

## Explanation

Machine learning models usually cannot interpret emojis directly.

The `emoji.demojize()` function converts emojis into readable text.

### Option 1: General NLP Tasks (Recommended)

For RAG, embeddings, semantic search, chunking, document retrieval, and question answering, emojis generally do not contribute meaningful information. Remove them using:

```python
import emoji

text = emoji.replace_emoji(text, replace="")
```

Example

```
I love Python 😊🚀
```

becomes

```
I love Python
```

### Option 2: Sentiment Analysis

For sentiment analysis, emojis contain emotional information and should be preserved by converting them into descriptive text.

```python
import emoji

text = emoji.demojize(text)
```

Example

```
😊
```

becomes

```
:smiling_face_with_smiling_eyes:
```

# Step 3: SpaCy Lemmatization

## Explanation

Lemmatization converts words to their base (dictionary) form while preserving their meaning.

Examples:

| Original Word | Lemma |
|---------------|-------|
| running | run |
| studies | study |
| went | go |
| children | child |
| cars | car |

Unlike stemming, lemmatization produces valid dictionary words.

SpaCy provides accurate linguistic lemmatization.

---

# Step 4: Recursive Chunking

## Explanation

Large Language Models have context length limitations.

Instead of sending the entire document, it is divided into smaller pieces called **chunks**.

Recursive Character Text Splitting divides text intelligently by checking separators in the following order:

1. Paragraph (`\n\n`)
2. Line (`\n`)
3. Space (` `)
4. Character (`""`)

This preserves sentence and paragraph boundaries as much as possible.

---

## Advantages

- Maintains context
- Produces meaningful chunks
- Better retrieval performance
- Ideal for RAG systems
- Prevents context window overflow

---

## Install LangChain

```python
pip install langchain
```

---

## Python Code

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = splitter.split_text(lemmatized_text)
```

---

# Step 5: Print the List of Chunks

## Explanation

After recursive chunking, the text is stored as a Python list.

Each element represents one chunk.

These chunks are later used for:

- Embedding generation
- Vector databases
- Semantic search
- Retrieval-Augmented Generation (RAG)

---

## Python Code

```python
for i, chunk in enumerate(chunks, start=1):
    print(f"Chunk {i}")
    print(chunk)
    print("-" * 80)
```

# Complete Pipeline

```python
import string
import emoji
import contractions
import spacy

from nltk.tokenize import word_tokenize
from langchain.text_splitter import RecursiveCharacterTextSplitter

# Load document
with open("sample.txt", "r", encoding="utf-8") as file:
    text = file.read()

# Lowercase
text = text.lower()

# Remove punctuation
text = ''.join(char for char in text if char not in string.punctuation)

# Remove special characters
text = ''.join(char if char.isalnum() or char.isspace() else ' ' for char in text)

# Handle emojis
text = emoji.demojize(text)

# Expand contractions
text = contractions.fix(text)

# Remove extra spaces
text = ' '.join(text.split())

# Tokenization
tokens = word_tokenize(text)

# Combine tokens
normalized_text = " ".join(tokens)

# Lemmatization
nlp = spacy.load("en_core_web_sm")
doc = nlp(normalized_text)
lemmatized_text = " ".join(token.lemma_ for token in doc)

# Recursive Chunking
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = splitter.split_text(lemmatized_text)

# Print chunks
for i, chunk in enumerate(chunks, start=1):
    print(f"Chunk {i}")
    print(chunk)
    print("-" * 80)
```
